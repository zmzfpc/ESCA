```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm
from mpl_toolkits.mplot3d import Axes3D

def plot_distribution(data, title="Data Distribution", save_name='', factor=1.1):
    data = np.array(data)
    abs_data = np.abs(data)  # Absolute value of the data
    max_value = abs_data.max()
    max_idx = np.unravel_index(np.argmax(abs_data), abs_data.shape)  # Find the index of the maximum value

    original_max_value = data[max_idx]  # Keep the original value for labeling

    tokens, channels = data.shape

    abs_values = abs_data.flatten()

    x, y = np.meshgrid(np.arange(tokens), np.arange(channels), indexing='ij')
    x = x.flatten()
    y = y.flatten()
    z = np.zeros_like(x)
    dz = abs_values

    norm = plt.Normalize(vmin=dz.min(), vmax=dz.max() * factor)
    cmap = cm.Reds
    colors = cmap(norm(dz))

    # Create a 3D plot
    fig = plt.figure(figsize=(8, 6))  # set figure size
    ax = fig.add_subplot(111, projection='3d')

    # adjust the bar width and depth
    ax.bar3d(x, y, z, dx=0.8, dy=0.8, dz=dz, color=colors, alpha=0.8)

    # adjust z limits
    ax.set_zlim(0, dz.max() * factor)

    # label the maximum value
    ax.text(max_idx[0], max_idx[1], max_value * 1.1,  # adjust the position of the text
            f"Max: {original_max_value:.6g}\n({max_idx[0]}, {max_idx[1]})",
            color='blue', fontsize=10, ha='center')
    ax.scatter(max_idx[0], max_idx[1], max_value, color='blue', s=50, label="Max Value")  # add a point for the max value

    ax.set_xlabel("X")
    ax.set_ylabel("Y")
    ax.set_zlabel("Value")
    ax.set_title(f"{title}")

    ax.legend(loc='upper right', bbox_to_anchor=(1.2, 1.0))  # adjust legend position
    
    if save_name:
        plt.savefig(save_name, dpi=300, bbox_inches='tight')
    else:
        plt.show()
    
    plt.close()
```

```python
def plot_qkv_distribution(data, title="QKV Distribution", save_name='', factor=1.1):
    data = np.array(data)
    
    # check if the data has 2304 in its shape and split accordingly
    if 2304 in data.shape:
        split_dim = data.shape.index(2304)
        if split_dim == 0:
            q_data, k_data, v_data = np.split(data, 3, axis=0)  # split along the first axis
        elif split_dim == 1:
            q_data, k_data, v_data = np.split(data, 3, axis=1)  # split along the second axis
        else:
            raise ValueError("Data dimensions are not compatible with QKV splitting.")
    else:
        raise ValueError("Input data must have one dimension equal to 2304.")
    
    qkv_data = [q_data, k_data, v_data]
    titles = ["Q Distribution", "K Distribution", "V Distribution"]

    # create a 3D plot
    fig, axes = plt.subplots(1, 3, figsize=(18, 6), subplot_kw={'projection': '3d'})

    for ax, sub_data, sub_title in zip(axes, qkv_data, titles):
        abs_data = np.abs(sub_data)
        max_value = abs_data.max()
        max_idx = np.unravel_index(np.argmax(abs_data), abs_data.shape)
        original_max_value = sub_data[max_idx]

        tokens, channels = sub_data.shape
        abs_values = abs_data.flatten()

        # create a meshgrid for the 3D plot
        x, y = np.meshgrid(np.arange(tokens), np.arange(channels), indexing='ij')
        x = x.flatten()
        y = y.flatten()
        z = np.zeros_like(x)
        dz = abs_values

        # color mapping
        norm = plt.Normalize(vmin=dz.min(), vmax=dz.max() * factor)
        cmap = cm.Reds
        colors = cmap(norm(dz))

        # generate the 3D bar plot
        ax.bar3d(x, y, z, dx=0.8, dy=0.8, dz=dz, color=colors, alpha=0.8)
        ax.set_zlim(0, dz.max() * factor)

        # label the maximum value
        ax.text(max_idx[0], max_idx[1], max_value * 1.1,
                f"Max: {original_max_value:.6g}\n({max_idx[0]}, {max_idx[1]})",
                color='blue', fontsize=10, ha='center')
        ax.scatter(max_idx[0], max_idx[1], max_value, color='blue', s=50, label="Max Value")

        # set axis labels and title
        ax.set_xlabel("X")
        ax.set_ylabel("Y")
        ax.set_zlabel("Value")
        ax.set_title(sub_title)
        ax.legend(loc='upper right')

    # set the main title for the figure
    fig.suptitle(title)
    if save_name:
        plt.savefig(save_name, dpi=300, bbox_inches='tight')
    else:
        plt.show()
    plt.close()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define a function to plot bias with outliers
def plot_bias_with_outliers(bias, save_path):
    """
    可视化 bias 数据，并标注出异常值 (outliers)。
    
    :param bias: np.ndarray, 提取的 bias 数据
    :param save_path: str, 保存绘图的路径
    """
    # calculate mean and standard deviation
    mean = np.mean(bias)
    std = np.std(bias)
    outliers = np.where(np.abs((bias - mean) / std) > 2)[0]  # identify outliers > 2 standard deviations
    
    # create a bar plot
    fig, ax = plt.subplots(figsize=(12, 6))  # set figure size
    ax.bar(range(len(bias)), bias, color='blue', alpha=0.7)
    ax.scatter(outliers, bias[outliers], color='red', label='Outliers', zorder=10)

    # add text annotations for outliers
    for idx in outliers:
        ax.text(idx, bias[idx] + np.max(bias) * 0.02, 
                f'{idx}', color='red', fontsize=10, ha='center', va='bottom')

    # set title and labels
    ax.set_title("Bias Visualization with Outliers Highlighted")
    ax.set_xlabel("Index")
    ax.set_ylabel("Bias Value")
    ax.set_ylim([np.min(bias) - np.std(bias), np.max(bias) + np.std(bias)])
    ax.legend()

    # save the plot
    plt.tight_layout()
    plt.savefig(save_path)
    plt.close()
    print(f"Saved bias visualization to {save_path}")

# Call the function with the bias data and save path
plot_bias_with_outliers(bias, 'bias_visualization_b12_norm1_epoch103.png')
```
